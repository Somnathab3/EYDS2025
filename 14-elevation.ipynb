```python
import os
import pandas as pd
import pyhigh
from pyhigh import get_elevation_batch

# Set base directory for datasets
base_dir = r"/kaggle/input/eyds-base-dataset"
sub_dir= r"/kaggle/working/"
# --- Create Elevation CSV for Training Data ---
train_file = os.path.join(base_dir, "Training_data.csv")
train_df = pd.read_csv(train_file)
# Get unique coordinates
train_coords = train_df[['Latitude', 'Longitude']].drop_duplicates()
# Convert to list of (lat, lon) tuples
train_coords_list = list(train_coords.itertuples(index=False, name=None))
# Retrieve elevations
train_coords['elevation'] = get_elevation_batch(train_coords_list)
```

```python
train_elev_file = os.path.join(sub_dir, "Training_set_elevation.csv")
train_coords.to_csv(train_elev_file, index=False)
print(f"Validation elevation data saved to {train_elev_file}")
```

```python
# Save to CSV
train_elev_file = os.path.join(sub_dir, "Training_set_elevation.csv")
train_coords.to_csv(train_elev_file, index=False)
print(f"Training elevation data saved to {train_elev_file}")
```

```python
# --- Create Elevation CSV for Validation Data ---
validation_file = os.path.join(base_dir, "Validation_data.csv")
validation_df = pd.read_csv(validation_file)
# Get unique coordinates
val_coords = validation_df[['Latitude', 'Longitude']].drop_duplicates()
# Convert to list of (lat, lon) tuples
val_coords_list = list(val_coords.itertuples(index=False, name=None))
# Retrieve elevations
val_coords['elevation'] = get_elevation_batch(val_coords_list)
# Save to CSV
val_elev_file = os.path.join(sub_dir, "validation_set_elevation.csv")
val_coords.to_csv(val_elev_file, index=False)
print(f"Validation elevation data saved to {val_elev_file}")
```
